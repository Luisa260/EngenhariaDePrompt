```python
# Solicita a temperatura do sistema ao usuário
temperature_input = input("Por favor, insira a temperatura atual do sistema em °C: ")

try:
    temperature = float(temperature_input)
    # Check if the temperature is greater than 30
    if temperature > 30:
        print(f"ALERTA: Superaquecimento! A temperatura atual é de {temperature}°C.")
    else:
        print(f"A temperatura atual é de {temperature}°C. Tudo normal.")
except ValueError:
    print("Entrada inválida. Por favor, insira um número para a temperatura.")
```

Criar um programa que organize arquivos bagunçados em pastas
automaticamente, baseando-se apenas na extensão do arquivo (ex: . pdf, .jpg).
Utilize biblioteca OS ou shutil.

```python
import os

# Define o diretório para os arquivos de teste
test_directory = "/content/minha_pasta_baguncada"

# Cria o diretório de teste se ele não existir
if not os.path.exists(test_directory):
    os.makedirs(test_directory)
    print(f"Diretório '{test_directory}' criado para testes.")
else:
    print(f"Diretório '{test_directory}' já existe. Certifique-se de que está limpo para um novo teste.")

# Lista de arquivos de teste com diferentes extensões
test_files = [
    "documento1.pdf", "documento2.docx", "relatorio.txt",
    "foto1.jpg", "imagem2.png", "logo.gif",
    "video_ferias.mp4", "tutorial.avi",
    "musica.mp3", "audio_aula.wav",
    "dados.zip", "backup.rar",
    "programa.exe",
    "script.py", "index.html",
    "desconhecido.xyz", "outro.dat", "temp.tmp"
]

# Cria os arquivos de teste vazios no diretório
for filename in test_files:
    file_path = os.path.join(test_directory, filename)
    with open(file_path, 'w') as f:
        f.write(f"Conteúdo de teste para {filename}")
    print(f"Arquivo '{filename}' criado em '{test_directory}'")

print("15 arquivos de teste criados com sucesso na pasta 'minha_pasta_baguncada'.")
```
